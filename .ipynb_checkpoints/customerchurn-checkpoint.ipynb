Predict behavior to retain customers. Analyze all relevant customer data and develop focused customer retention programs." [IBM Sample Data Sets]

Content:-
Each row represents a customer, each column contains customer’s attributes described on the column Metadata.

The data set includes information about:

Customers who left within the last month – the column is called Churn
Services that each customer has signed up for – phone, multiple lines, internet, online security, online backup, device protection, tech support, and streaming TV and movies
Customer account information – how long they’ve been a customer, contract, payment method, paperless billing, monthly charges, and total charges
Demographic info about customers – gender, age range, and if they have partners and dependents

Data Cleaning->

```python
df['TotalCharges'].describe()
df['TotalCharges'].hist()
```

as the following distribution is skewed we will add median values to the missing data 

```python
df['TotalCharges']=pd.to_numeric(df['TotalCharges'],errors='coerce')
df['TotalCharges'].dtype
```

```python
df['TotalCharges'].isnull().sum()
```

```python
df['TotalCharges']=df['TotalCharges'].fillna(df['TotalCharges'].median())
```

```python
df['TotalCharges'].isnull().sum()
```

```python
df=df.drop(columns=['customerID','Churn'])
```

now we will proceed to model the dataset based on which we can make predictions and understand which type  of customers are leaving

```python
df.head()
```

```python
x = df
```

```python
y = df_raw['Churn'].map({'No': 0, 'Yes': 1})
```

```python
x.head()
```

```python
y.head()
```

now we will split the data for training and testing 

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=SEED,stratify=y)
```

```python
print(f"train:{x_train.shape},test:{x_test.shape}")
```

now we will scale the data 
Why Scaling?
Scaling is used to bring all numerical features to a similar range.
This prevents columns with large values from dominating smaller-value columns.
It helps models like Logistic Regression, KNN, SVM, and Neural Networks learn faster and more accurately.
Scaling also improves convergence and reduces training time.
Common methods: StandardScaler, MinMaxScaler.
usually if not scaled error like this comes STOP: TOTAL NO. OF ITERATIONS REACHED LIMIT

Increase the number of iterations to improve the convergence (max_iter=100). 

as the difference in values of column 1 and column 2 is very large so model has to take more iterations ( compute more ) or more time to learn, One small number, one huge number.Model struggles to balance both. 
hence we always check wether scaling needs to be done or not usually except decision trees , xgboost and random forest we always scale the data 
Default tries are too low
Model gets only 100 chances by default in logistic regression

```python
num=x_train.select_dtypes(include=['int64','float']).columns.tolist()
cat=x_train.select_dtypes(include=['object','category']).columns.tolist()
preprocessor=ColumnTransformer(transformers=[('num',StandardScaler(),num),
                               ('cat',OneHotEncoder(drop='first',handle_unknown='ignore'),cat)])
```

```python
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=SEED)
```

```python
m1pipeline=Pipeline(steps=[('preprocessor',preprocessor),
                         ('model',LogisticRegression(max_iter=2000,random_state=SEED))])
                
```

```python
m1pipeline.fit(x_train,y_train)
ypred1=m1pipeline.predict(x_test)
```

```python
print(f"Baseline accuracy of logistic regression:{accuracy_score(y_test, ypred1)}")
print(classification_report(y_test, ypred1))
```

```python
cv_score1=cross_val_score(m1pipeline,x_train,y_train,cv=cv,scoring='roc_auc')
print(cv_score1)
```

```python
print(f"baseline logistic roc_auc:{cv_score1.mean():.3f}±{cv_score1.std():.3f}")
```

```python
m2pipeline=Pipeline(steps=[('preprocessor',preprocessor),
                           ('model',LogisticRegression(class_weight='balanced',max_iter=2000,random_state=SEED))])
```

```python
m2pipeline.fit(x_train,y_train)
ypred2=m2pipeline.predict(x_test)
```

```python
print(f"Balanced accuracy: {accuracy_score(y_test, ypred2)}")
print(classification_report(y_test, ypred2))
```

```python
cv_score2=cross_val_score(m2pipeline,x_train,y_train,cv=cv,scoring='roc_auc')
print(cv_score2)
```

```python
print(f"balanced logistic roc_auc:{cv_score2.mean():.3f}±{cv_score2.std():.3f}")
```

this classification report tells us ->
Precision for churn = when model says churn, how often right?
Recall for churn = how many actual churners it caught
F1-score = balance of both

```python
m3pipeline=Pipeline(steps=[('preprocessor',preprocessor),
                           ('model',XGBClassifier(random_state=SEED,eval_metric='logloss'))])
```

```python
m3pipeline.fit(x_train,y_train)
ypred3=m3pipeline.predict(x_test)
```

```python
print(f"xgb baseline accuracy:{accuracy_score(y_test,ypred3)}")
print(classification_report(y_test,ypred3))
```

```python
cv_score3=cross_val_score(m3pipeline,x_train,y_train,cv=cv,scoring='roc_auc')
print(cv_score3)
```

```python
print(f"baseline xgb roc_auc:{cv_score3.mean():.3f}±{cv_score3.std():.3f}")
```

```python
param_dist = {
    'model__n_estimators':      [100, 150, 200, 300],
    'model__max_depth':         stats.randint(3, 8),
    'model__learning_rate':     stats.uniform(0.01, 0.19),
    'model__subsample':         stats.uniform(0.6, 0.4),
    'model__scale_pos_weight':  [1, 2, 3, 4]   # ratio ≈ negatives/positives ≈ 2.8
}
search = RandomizedSearchCV(
    m3pipeline, param_dist,
    n_iter=30, cv=cv, scoring='roc_auc',
    random_state=SEED, n_jobs=-1, verbose=1
)
search.fit(x_train, y_train)
ypred4=search.predict(x_test)
print(f"Best CV ROC-AUC: {search.best_score_:.3f}")
print(f"Best params: {search.best_params_}")
xgb_best = search.best_estimator_
```

```python
print(f"tuned xgb accuracy:{accuracy_score(ypred4,y_test)}")
print(classification_report(ypred4,y_test))
```

```python
from sklearn.metrics import RocCurveDisplay, PrecisionRecallDisplay, ConfusionMatrixDisplay

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

RocCurveDisplay.from_estimator(m2pipeline, x_test, y_test, ax=axes[0], name='Logistic (balanced)')
RocCurveDisplay.from_estimator(xgb_best, x_test, y_test, ax=axes[0], name='XGBoost (tuned)')
axes[0].plot([0, 1], [0, 1], linestyle='--', color='gray', label='Random guess')
axes[0].set_title('ROC curve — test set')
axes[0].legend()

PrecisionRecallDisplay.from_estimator(m2pipeline, x_test, y_test, ax=axes[1], name='Logistic (balanced)')
PrecisionRecallDisplay.from_estimator(xgb_best, x_test, y_test, ax=axes[1], name='XGBoost (tuned)')
axes[1].set_title('Precision-Recall curve — test set')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
ConfusionMatrixDisplay.from_estimator(m2pipeline, x_test, y_test, ax=axes[0], cmap='Blues')
axes[0].set_title('Logistic (balanced)')
ConfusionMatrixDisplay.from_estimator(xgb_best, x_test, y_test, ax=axes[1], cmap='Blues')
axes[1].set_title('XGBoost (tuned)')
plt.tight_layout()
plt.show()
```

```python
feature_names = m2pipeline.named_steps['preprocessor'].get_feature_names_out()
coefficients  = m2pipeline.named_steps['model'].coef_[0]

coeff_df = pd.DataFrame({'feature': feature_names, 'coefficient': coefficients})
coeff_df = coeff_df.sort_values('coefficient', ascending=False)   # this one actually runs

print("Top 5 features pushing toward churn:")
print(coeff_df.head())
print("\nTop 5 features pushing toward staying:")
print(coeff_df.tail())
```

```python
import shap

# SHAP's TreeExplainer needs the raw model + already-preprocessed data —
# not the full Pipeline object, since Pipeline bundles preprocessing + model together
X_test_transformed = xgb_best.named_steps['preprocessor'].transform(x_test)
feature_names_xgb   = xgb_best.named_steps['preprocessor'].get_feature_names_out()

explainer   = shap.TreeExplainer(xgb_best.named_steps['model'])
shap_values = explainer(X_test_transformed)
shap_values.feature_names = list(feature_names_xgb)

shap.summary_plot(shap_values, X_test_transformed, feature_names=feature_names_xgb)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score

y_pred_log = m2pipeline.predict(x_test)
y_prob_log = m2pipeline.predict_proba(x_test)[:, 1]

y_pred_xgb = xgb_best.predict(x_test)
y_prob_xgb = xgb_best.predict_proba(x_test)[:, 1]

comparison = pd.DataFrame({
    'Model':        ['Logistic (balanced)', 'XGBoost (tuned)'],
    'Accuracy':     [accuracy_score(y_test, y_pred_log),  accuracy_score(y_test, y_pred_xgb)],
    'Recall (1)':   [recall_score(y_test, y_pred_log),    recall_score(y_test, y_pred_xgb)],
    'Precision (1)':[precision_score(y_test, y_pred_log), precision_score(y_test, y_pred_xgb)],
    'F1 (1)':       [f1_score(y_test, y_pred_log),        f1_score(y_test, y_pred_xgb)],
    'ROC-AUC':      [roc_auc_score(y_test, y_prob_log),   roc_auc_score(y_test, y_prob_xgb)],
}).round(3)

display(comparison)
```
